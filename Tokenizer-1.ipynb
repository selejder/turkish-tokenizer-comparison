```python
def get_stats(ids):

    stats = {}
    
    for pair in zip(ids, ids[1:]):
        if pair in stats:
            stats[pair] += 1
        else:
            stats[pair] = 1

    return stats
```

```python
def merge(ids, pair, idx):

    new_ids = []

    ix = 0
    while ix < len(ids):
        
        if pair[0] == ids[ix] and ix < len(ids) - 1 and pair[1] == ids[ix+1]:
            new_ids.append(idx)
            ix += 2

        else:
            new_ids.append(ids[ix])
            ix += 1
    return new_ids
```

```python
encoded_text = new_data.encode('utf-8')
print(f"Encoded Text: {list(encoded_text[:200])}\n\n")

stats = get_stats(encoded_text)
#print(f"Stats: {stats}\n\n")

target_pair = max(stats, key=stats.get)
print(f"Target Pair: {target_pair}\n\n")

merged_text = merge(encoded_text, target_pair, 256)
print(f"Merged Text: {list(merged_text[:100])}")
```

```python
def train_bpe(ids, num_merge):
    merges = {}
    idx = 256
    for i in range(num_merge):
        stats = get_stats(ids)
        target_pair = max(stats, key=stats.get)
        ids = merge(ids, target_pair, idx)
        merges[target_pair] = idx
        idx += 1

    return ids, merges
```

```python
def encoder(text, merges):
    encoded_ids = text.encode('utf-8')

    for pair, token_id in merges.items():
        encoded_ids = merge(encoded_ids, pair, token_id)

    return encoded_ids
```

```python
def decoder(encoded_ids, vocab):
    text = []
    for idx in encoded_ids:
        ch = vocab[idx]
        text.append(ch)

    text = b''.join(text)
    text = text.decode('utf-8')
    
    return text
```

```python
def batcher(encoded_ids, block_size: int = 256, batch_size: int = 32):
    X = []
    targets = []

    for i in range(batch_size):
        ix = torch.randint(0, len(encoded_ids) - (block_size + 1), (1, ))

        X.append(encoded_ids[ix:ix+block_size])
        targets.append(encoded_ids[ix+1:ix+block_size+1])

    X = torch.tensor(X)
    targets = torch.tensor(targets)

    return X, targets
```

```python
class EmbeddingOperation(nn.Module):
    def __init__(self, vocab_size: int = 211, block_size: int = 256, embed_dim: int = 256):
        super().__init__()
        
        self.token_embedding = nn.Embedding(vocab_size, embed_dim)
        self.position_embedding = nn.Embedding(block_size, embed_dim)

    def forward(self, ix):
        B, T = ix.shape
        self.tokens = self.token_embedding(ix)
        self.positions_ids = torch.arange(0, T, device = ix.device)
        self.positions = self.position_embedding(self.positions_ids)

        output = self.tokens + self.positions

        return output
```

```python
class SingleHeadAttention(nn.Module):
    def __init__(self,  head_size: int = head_size, embed_dim: int = embed_dim, block_size: int = block_size):
        super().__init__()
        self.head_size = head_size
        
        self.q_proj = nn.Linear(in_features=embed_dim, out_features=head_size, bias=False)
        self.k_proj = nn.Linear(in_features=embed_dim, out_features=head_size, bias=False)
        self.v_proj = nn.Linear(in_features=embed_dim, out_features=head_size, bias=False)
        mask = torch.tril(torch.ones(block_size, block_size))
        self.register_buffer('mask', mask)
        
    def forward(self, x):
        B, T, C = x.shape
        
        Q = self.q_proj(x) # Q -> [batch_size, block_size, head_size]
        K = self.k_proj(x) # K -> [batch_size, block_size, head_size]
        V = self.v_proj(x) # V -> [batch_size, block_size, head_size]
        K_t = K.transpose(-2, -1) # K_t -> [batch_size, head_size, block_size]
        wei = Q @ K_t # wei -> [batch_size, block_size, block_size]
        
        scale = self.head_size**0.5 
        scaled_score = wei / scale #scaled_score -> [batch_size, block_size, block_size]
        
        casual_mask = self.mask[:T, :T]
        forbidden = casual_mask == 0
        masked_score = scaled_score.masked_fill(
            forbidden,
            float("-inf")
        )
        
        attention_weights = torch.softmax(masked_score, dim = -1)

        output = attention_weights @ V # output -> [batch_size, head_size, block_size]

        return output
```

```python
class MultiHeadAttention(nn.Module):
    def __init__(self, num_head: int = num_head, embed_dim: int = embed_dim, head_size: int = head_size):
        super().__init__()
        self.heads = nn.ModuleList([
            SingleHeadAttention()
            for _ in range(num_head)
        ]) # we are creating the multiheadattention which encompasse num_head single head attention 

        self.out_projection = nn.Linear(in_features=num_head*head_size, out_features=embed_dim) #

    def forward(self, x):
        head_outputs = [head(x) for head in self.heads]

        head_outputs = torch.cat(head_outputs, dim = -1)
    
        output = self.out_projection(head_outputs)
        
        return output
```

```python
class FeedForwardNetwork(nn.Module):
    def __init__(self, embed_dim: int = embed_dim):
        super().__init__()

        self.layers = nn.Sequential(
            nn.Linear(in_features=embed_dim, out_features=4*embed_dim),
            nn.ReLU(),
            nn.Linear(in_features=embed_dim*4, out_features=embed_dim*4),
            nn.ReLU(),
            nn.Linear(in_features=embed_dim*4, out_features=embed_dim)
        )

    def forward(self, x):
        
        output = self.layers(x)
        
        return output
```

```python
class MyTransformer(nn.Module):
    def __init__(self, embed_dim: int = embed_dim):
        super().__init__()

        self.ln1 = nn.LayerNorm(normalized_shape = embed_dim)
        self.ln2 = nn.LayerNorm(normalized_shape = embed_dim)
        self.MHA = MultiHeadAttention()
        self.ffn = FeedForwardNetwork()
        
    def forward(self, x):
        B, T, C = x.shape
        x = self.MHA(self.ln1(x)) + x
        x = self.ffn(self.ln2(x)) + x

        output = x

        return output
```

```python
class GPT(nn.Module):
    def __init__(self, 
                 block_size: int = block_size,
                 batch_size: int = batch_size,
                 num_head: int = num_head,
                 vocab_size: int = vocab_size,
                 embed_dim: int = embed_dim,
                 head_size: int = head_size,
                 num_transformer: int = num_transformer
                ):
        super().__init__()

        self.embedding_layer = EmbeddingOperation(
            vocab_size=vocab_size,
            block_size=block_size,
            embed_dim=embed_dim
        )
        self.transfomer = MyTransformer(
                embed_dim=embed_dim
            )

    def forward(self, x):
        x = self.embedding_layer(x)
        x = self.transfomer(x)
        
        output = x
        
        return output
```
